```python
a = input("nhap ho ten chu ho:")
b= int(input("nhap so dien thang truoc: "))
c= int(input("nhap so dien thang nay: "))
T =c-b
if T <= 50:
    S = (T*1984)*(108/100)      
    print("ten chu ho: ",a)
    print("so dien thang truoc: ",b)
    print("so dien thang nay: ",c)
    print("so tien dien phai tra la: ",S)
elif T > 50 and T <= 100:
    S = (50*1984 + (T-50)*2050)*(108/100)      
    print("ten chu ho: ",a)
    print("so dien thang truoc: ",b)
    print("so dien thang nay: ",c)
    print("so tien dien phai tra la: ",S)
elif T > 100 and T <= 200:
    S = (50*1984 + 50*2050 + (T-100)*2380)*(108/100)      
    print("ten chu ho: ",a)
    print("so dien thang truoc: ",b)
    print("so dien thang nay: ",c)
    print("so tien dien phai tra la: ",S)
elif T > 200 and T <= 300:
    S = (50*1984 + 50*2050 + 100*2380 + (T-200)*2998)*(108/100)      
    print("ten chu ho: ",a)
    print("so dien thang truoc: ",b)
    print("so dien thang nay: ",c)
    print("so tien dien phai tra la: ",S)
elif T > 300 and T <= 400:
    S = (50*1984 + 50*2050 + 100*2380 + 100*2998 + (T-300)*3350)*(108/100)      
    print("ten chu ho: ",a)
    print("so dien thang truoc: ",b)
    print("so dien thang nay: ",c)
    print("so tien dien phai tra la: ",S)
elif T > 400:
    S = (50*1984 + 50*2050 + 100*2380 + 100*2998 + 100*3350 + (T-400)*3460)*(108/100)      
    print("ten chu ho: ",a)
    print("so dien thang truoc: ",b)
    print("so dien thang nay: ",c)
    print("so tien dien phai tra la: ",S)
```
